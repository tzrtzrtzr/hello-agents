### 创建虚拟环境

#### **虚拟环境的好处** ：  
- **避免依赖包版本冲突**  
比如你项目 A 需要 requests==2.25.1，项目 B 需要 requests==2.31.0。如果都安装在系统全局 Python 中，只能留一个版本，另一个项目就会报错。虚拟环境会给每个项目单独安装一套依赖，互不干扰。 

- **不污染系统全局环境**  
很多第三方包会修改系统级的 Python 配置，或者安装大量依赖。虚拟环境里的包不会影响系统 Python，也不会干扰其他项目，出问题直接删掉虚拟环境即可，不用重装 Python。 

- **项目依赖可精准复现**  
虚拟环境中可以通过 pip freeze > requirements.txt 导出当前项目的所有依赖和版本。其他人拿到你的项目，用这个文件就能一键安装完全相同的依赖，避免 “我本地能跑，你那跑不了” 的环境差异问题。

- **权限与安全隔离**  
系统级 Python 安装包需要管理员权限，虚拟环境里的操作完全在项目目录内，不需要管理员权限，也不会修改系统文件，更安全。
